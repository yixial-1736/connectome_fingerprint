```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
import pandas as pd
import seaborn as sns
```

# Permutation Test function

```python
def individual_permutation_test(prediciton,truth, n_permutations=1000):
    """
    Perform individual permutation tests for each feature in the dataset.
    
    Parameters:
    prediction (csv file): 1D array of predicted activations based on ridge model
    truth (csv file): 1D array of true activations
    n_permutations (int): Number of permutations to perform.
    
    Returns:
    p_value (float): The p-value that the observed correlation is significant.
    permuted_corrs (numpy.ndarray): 1-D array of correlation coefficients from each permutation.
    """

    # Calculate the observed correlation
    observed_corr, _ = pearsonr(prediciton, truth)

    # permutation distribution
    permuted_corrs = np.zeros(n_permutations)
    for i in range(n_permutations):
        permuted_truth = np.random.permutation(prediciton)
        permuted_corrs[i], _ = pearsonr(permuted_truth, truth)

    # Calculate the p-value
    # one-tailed test
    p_value = np.sum(permuted_corrs >= observed_corr) / n_permutations
    return p_value, permuted_corrs

def plot_permutation_distribution(permuted_corrs, observed_corr):
    """
    Plot the distribution of permuted correlation coefficients and the observed correlation.
    
    Parameters:
    permuted_corrs (numpy.ndarray): 1-D array of correlation coefficients from each permutation.
    observed_corr (float): The observed correlation coefficient.
    """
    plt.hist(permuted_corrs, bins=50, color='gray', alpha=0.8, label='Null Distribution (Shuffled)')
    plt.axvline(observed_corr, color='red', linestyle='--', linewidth=2, label=f'True Correlation = {observed_corr:.3f}')
    plt.title('Permutation Test Distribution')
    plt.xlabel('Correlation Coefficient')
    plt.ylabel('Frequency')
    plt.legend()
    plt.show()
    
```

# Example Usage

- This is a quick demo on the test
- Note: this is for single subject

```python
# Done at the subject level & the group level
subjects = ['RPMS_2002.csv', 'RPMS_2003.csv', 'RPMS_2004.csv', 'RPMS_2005.csv', 'RPMS_2006.csv', 'RPMS_2007.csv', 'RPMS_2009.csv', 'RPMS_2010.csv', 'RPMS_2011.csv', 'RPMS_2012.csv', 'RPMS_2013.csv', 'RPMS_2014.csv', 'RPMS_2015.csv', 'RPMS_2022.csv', 'RPMS_2024.csv', 'RPMS_2026.csv', 'RPMS_2027.csv', 'RPMS_2028.csv']

results = []
for subject in subjects:
    print({subject})
    predicted_data_right = pd.read_csv(f"/ConnectomeFingerprinting/Outputs/Ridge_Results/hypo1/SymR-vs-SymU/rh_Sym_pred_{subject}",header=None)
    true_data_right = pd.read_csv(f"/ConnectomeFingerprinting/Outputs/Scan_Activations/hypo1/SymR-vs-SymU/rh.{subject}", sep=",", header=None)
    true_data_right = true_data_right.values.flatten()
    predicted_data_right = predicted_data_right.values.flatten()
    predicted_data_left = pd.read_csv(f"/ConnectomeFingerprinting/Outputs/Ridge_Results/hypo1/SymR-vs-SymU/lh_Sym_pred_{subject}", header=None)
    true_data_left = pd.read_csv(f"/ConnectomeFingerprinting/Outputs/Scan_Activations/hypo1/SymR-vs-SymU/lh.{subject}", sep=",", header=None)
    true_data_left = true_data_left.values.flatten()
    predicted_data_left = predicted_data_left.values.flatten()
    
    print("Predicted Right shape:", predicted_data_right.shape)
    print("True Right shape:", true_data_right.shape)
    print("Predicted Left shape:", predicted_data_left.shape)
    print("True Left shape:", true_data_left.shape)
    
    # Perform permutation test
    p_value_r, permuted_corrs_r = individual_permutation_test(predicted_data_right, true_data_right, n_permutations=1000)
    p_value_l, permuted_corrs_l = individual_permutation_test(predicted_data_left, true_data_left, n_permutations=1000)
    
    observed_corr_r, _ = pearsonr(predicted_data_right, true_data_right)
    observed_corr_l, _ = pearsonr(predicted_data_left, true_data_left)

    # Summaries for plotting
    mean_null_r = np.mean(permuted_corrs_r)
    lower_ci_r = np.percentile(permuted_corrs_r, 2.5)
    upper_ci_r = np.percentile(permuted_corrs_r, 97.5)

    mean_null_l = np.mean(permuted_corrs_l)
    lower_ci_l = np.percentile(permuted_corrs_l, 2.5)
    upper_ci_l = np.percentile(permuted_corrs_l, 97.5)
    
    print(f'Model Prediction: {observed_corr_r}')
    print(f'Model Prediction: {observed_corr_l}')
    print(f'P-value of above chance-level model performance: {p_value_r:.4f}')
    print(f'P-value of above chance-level model performance: {p_value_l:.4f}')
    
    plot_permutation_distribution(permuted_corrs_r, observed_corr_r)
    plot_permutation_distribution(permuted_corrs_l, observed_corr_l)

    # Store in results list for group plotting
    results.append({
        "subject": subject,
        "hemi": "RH",
        "real_corr": observed_corr_r,
        "mean_null": mean_null_r,
        "lower_ci": lower_ci_r,
        "upper_ci": upper_ci_r,
        "p_value": p_value_r
    })
    results.append({
        "subject": subject,
        "hemi": "LH",
        "real_corr": observed_corr_l,
        "mean_null": mean_null_l,
        "lower_ci": lower_ci_l,
        "upper_ci": upper_ci_l,
        "p_value": p_value_l
    })


# Convert to DataFrame
results_df = pd.DataFrame(results)

# -------------------------------
# Visualization: Real vs Null per subject
# -------------------------------
plt.figure(figsize=(10, 5))

# Sort subjects for cleaner plotting
results_df = results_df.sort_values(by=["subject", "hemi"])
x_labels = [f"{s}_{h}" for s, h in zip(results_df["subject"], results_df["hemi"])]
x = np.arange(len(results_df))

# Plot error bars (null 95% CI)
plt.errorbar(x, results_df["mean_null"], 
             yerr=[results_df["mean_null"] - results_df["lower_ci"], 
                   results_df["upper_ci"] - results_df["mean_null"]],
             fmt='o', color='gray', ecolor='lightgray', elinewidth=2, capsize=3, label='Null (95% CI)')

# Overlay real correlations
plt.scatter(x, results_df["real_corr"], color='red', s=40, label='Real Correlation')

plt.xticks(x, x_labels, rotation=90)
plt.ylabel("Correlation (Predicted vs. Actual)")
plt.title("Real vs. Null Correlations per Subject (with 95% Null CI)")
plt.legend(frameon=False)
plt.tight_layout()
plt.show()

# Optional: print summary statistics
sig_count = np.sum(results_df["p_value"] < 0.05)
print(f"\n{sig_count}/{len(results_df)} hemisphere-level results are significant at p<0.05.")
```

```python
# Compute mean null per hemisphere
mean_nulls = results_df.groupby("hemi")["mean_null"].mean()

plt.figure(figsize=(6, 6))
sns.violinplot(
    data=results_df, 
    x="hemi", 
    y="real_corr", 
    inner="point",  # shows individual subjects as small points
    cut=0,
    linewidth=1,
    palette={"LH": "#7FDBB6", "RH": "#FFB07C"}
)

# Overlay mean null lines for each hemisphere
for hemi, mean_null in mean_nulls.items():
    plt.axhline(mean_null, color="black", linestyle="--", linewidth=1)
    plt.text(
        x={"LH": 0, "RH": 1}[hemi], 
        y=mean_null + 0.005, 
        s="Mean null", 
        ha="center", 
        va="bottom", 
        fontsize=9, 
        color="black"
    )

plt.title("Predicted Associative vs. Actual Correlations by Hemisphere")
plt.ylabel("Correlation Coefficient")
plt.xlabel("Hemisphere")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 5))
for hemi, color_real, color_null in [("LH", "red", "gray"), ("RH", "blue", "gray")]:
    mask = results_df["hemi"] == hemi
    xi = x[mask]
    # Null means + CI
    plt.errorbar(
        xi,
        results_df.loc[mask, "mean_null"],
        yerr=[
            results_df.loc[mask, "mean_null"] - results_df.loc[mask, "lower_ci"],
            results_df.loc[mask, "upper_ci"] - results_df.loc[mask, "mean_null"]
        ],
        fmt='o',
        color=color_null,
        ecolor='lightgray',
        elinewidth=1.5,
        capsize=3,
        markersize=6,
        zorder=2,
        label=f'Null (95% CI)' if hemi == "LH" else None  # avoid duplicate labels
    )
    # Real correlations
    plt.scatter(
        xi,
        results_df.loc[mask, "real_corr"],
        color=color_real,
        s=50,
        zorder=3,
        label=f'Real Correlation - {hemi}'
    )

plt.axhline(0.0, color='black', linewidth=0.8, alpha=0.25, zorder=0)
plt.xticks(x, x_labels, rotation=90)
plt.ylabel("Correlation (Predicted vs. Actual)")
plt.title("Real vs. Null Correlations per Subject (by Hemisphere)")
plt.legend(frameon=False)
plt.tight_layout()
plt.savefig("Real_v_Null_WholeBrain.tif", dpi=300, bbox_inches='tight')
plt.show()

```
